# Information Presentation & Data Visualisation

## 13 Lab Exercise sheet: State-of-the-art visualisations

In this lab, you will critically analyse a state-of-the-art scientific paper on data visualisations. Using the provided checklist, you will evaluate the paper’s methodology, findings, and contribution to the field. This task aims to develop your ability to assess the strengths and weaknesses of scientific work, with a focus on visualisation techniques, and to foster a deeper understanding of current trends and innovations in the field of data visualisation.

## Critical reading checklist:
Please use the following checklist to critique a state-of-the-art scientific paper on data visualisations of your choice.

- **Purpose:** What is the main question or goal?
- **Contribution:** What does the paper add that’s new?
- **Methodology:** Is the method appropriate and rigorous?
- **Evidence:** Are claims supported by data or citations?
- **Clarity:** Is the argument logical and well-structured?
- **Bias & Limitations:** What's missing or underexplored?
