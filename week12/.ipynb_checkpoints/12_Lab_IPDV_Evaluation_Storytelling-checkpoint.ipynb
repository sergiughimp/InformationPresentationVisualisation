# Information Presentation & Data Visualisation

## 12 Lab Exercise sheet: Evaluation and Storytelling

In this lab, you will evaluate data visualisation techniques applied to real-world datasets to effectively communicate insights. 

## Evaluation criteria:
Please use the following checklist to evaluate the visualisations:

- **Relevance:** Does the visualisation match the goals of the dataset? Are irrelevant details included?
- **Clarity:** Is the visualisation easy to understand? Are the axes, labels, and data points clear?
- **Design:** Are the colours, font size, and other design elements appropriate? Can any design be improved?
- **Efficiency:** How quickly and easily the visualisation allows users to extract insights and interpret the data?

## Task 1:Understanding the dataset
- Load a dataset of your choice and explore its structure.

## Task 2: Creating initial visualisations
- Generate at least two different types of charts to visualise key trends.

## Task 3: Evaluating the visualisations
- Evaluate relevance, clarity, design, and effectiveness

## Task 4: Improving the visualisations
- Modify charts based on evaluation to improve clarity and effectiveness.

## Task 5: Keep improving the visualisations
- Modify charts based on evaluation to improve clarity and effectiveness.

## Task 5: Communicating insights
- Write a short explanation of the key insights from the journey of improving your visualisations.
