# Information Presentation & Data Visualisation

## 05 Lab Exercise sheet: Interaction & Animation

This lab is designed to provide hands-on practice with interaction and animation concepts in data visualisation.

## Task 1: Understanding your audience

a) Create two distinct personas representing potential users of your data visualisation, considering their background, goals, and level of data literacy.

b) Use a Large Language Model (LLM) to generate two written interpretations of a data visualisation: one for each persona. One interpretation should present accurate information, while the other should convey misinformation.

c) Analyse and evaluate these interpretations, assessing their clarity, potential biases, and impact on different audiences. Discuss how each persona might perceive and respond to the information.

## Task 2: Interactive Scatter Plot

Create an interactive scatter plot of the cars dataset using Altair, filter by Origin and adjust the size of points based on Horsepower.

## Task 3: Animated Line Plot 

Create an animated line plot of stock prices over time. Animate the price of "IBM" over time in the `stocks` dataset.

## Task 4: Interactive Bar Chart 

Create a bar chart showing the number of movies per genre from the `movies` dataset. Add an interactive dropdown to select a specific genre.

## Task 5: Animated Histogram

Create an animated histogram of `Miles_per_Gallon` for cars, updating the bins dynamically.

## Task 6: Interactive Map 

Create an interactive map showing flight routes from the flights_5k and airports datasets and a dropdown menu where users can filter the routes by Origin
