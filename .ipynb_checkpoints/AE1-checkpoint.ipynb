### Taks 1: Critique of Figure 5 from Griffin & Mylonas (2019)

Figure 5 from *Categorical Colour Geometry* presents a clear visual summary of how people name colors across languages. The circular layout and use of each color’s average hue help viewers understand which color groups are most common globally. Visually, the figure conveys the idea that colors exist along a continuous spectrum rather than as separate categories. The size of each segment effectively shows naming frequency, making major color terms like blue, green, and red easy to identify across cultures.

While the design successfully highlights overall trends, its **effectiveness** weakens when viewers try to see smaller differences. Many sectors share similar brightness and saturation, which makes it hard to distinguish one color boundary from another. Because the figure has no numerical scale or detailed labels, it communicates general patterns well but lacks precision for direct data comparison.

From a design point of view, the visualization partly follows **Gestalt principles** such as similarity and grouping but could better reflect **human visual perception** by using stronger contrasts and clearer separation. The circular layout is aesthetically pleasing, yet it demands mental effort to compare segments that could be easier to interpret with a more linear or grid format.

To improve understanding, two practical changes are suggested:
1. Add a **secondary chart or small numerical labels** to show exact frequency counts.
2. **Enhance color contrast or add subtle boundaries** between sectors.

These refinements would preserve the figure’s visual appeal while making it clearer and more informative.


### Task 2: Data Preparation and Transformation for Colour Analysis

The purpose of this task was to **load, clean, and prepare** the raw colour naming dataset for analysis and presentation.
The dataset, collected from *colornaming.net* and published by Mylonas et al. (2017), contained both metadata and colour information that needed to be processed before further use.

#### Steps performed

1. **Load the raw dataset**: The CSV file was loaded including metadata and licensing information, and the first rows were inspected to understand its structure.
2. **Identify the header row**: The row containing the actual dataset column names (`target_id`, `name`, `country_resident`, `red`, `green`, `blue`) was located to skip metadata.
3. **Reload the dataset**: The CSV was reloaded skipping metadata rows, focusing only on the relevant data.
4. **Clean text columns**: All text columns were cleaned by converting to lowercase, removing special characters, and trimming extra spaces.
5. **Convert numeric columns**: The `red`, `green`, and `blue` columns were converted to numeric types to allow further analysis.
6. **Remove invalid or missing values**: Rows with missing or invalid RGB values were removed to ensure data integrity.
7. **Remove rare colour names**: Colour names appearing fewer than five times were filtered out to focus on common terms.
8. **Reset the index**: The dataset index was reset for an organized and clean structure.
9. **Save cleaned dataset**: The processed dataset was saved as `cleaned_CN_EN_10K_Mylonas2017.csv` for further use.
10. **Map to broader colour groups**: Detailed colour names were grouped into broader categories (e.g., red, blue, green) for easier analysis.
11. **Compute mean RGB values per group**: The mean RGB values for each colour group were calculated.
12. **Convert mean RGB to hex**: Mean RGB values were converted to hex codes for accurate visual representation.
13. **Create Altair visualization**: An interactive Altair scatter plot was created showing mean RGB values per colour group, with colors reflecting the true mean RGB and tooltips displaying group names and values.

This structured workflow ensured the dataset was **clean, consistent, and visually interpretable**, ready for analysis, visualization, or further applications.



```python
import pandas as pd
import re
import altair as alt

# --- Step 1: Load raw CSV (including metadata) ---
df_raw = pd.read_csv('CN_EN_10K_Mylonas2017.csv', header=None)

# --- Print initial raw data preview ---
print("📄 Initial raw data preview (first 10 rows):")
print(df_raw.head(10))
print("\n" + "-"*60 + "\n")

# --- Step 2: Identify the correct header row ---
header_row = df_raw[df_raw[0] == 'target_id'].index[0]

# --- Step 3: Reload dataset skipping metadata rows ---
df = pd.read_csv('CN_EN_10K_Mylonas2017.csv', skiprows=header_row)

# --- Step 4: Clean text columns ---
def clean_text(x):
    if isinstance(x, str):
        x = x.lower().strip()
        x = re.sub(r'[^a-z0-9\s]', '', x)
        x = re.sub(r'\s+', ' ', x)
    return x

df['name'] = df['name'].map(clean_text)

# --- Step 5: Convert RGB columns to numeric ---
for col in ['red', 'green', 'blue']:
    df[col] = pd.to_numeric(df[col], errors='coerce')

# --- Step 6: Remove invalid/missing RGB values ---
df = df.dropna(subset=['red', 'green', 'blue'])

# --- Step 7: Remove rare color names (<5 occurrences) ---
df = df[df['name'].map(df['name'].value_counts()) >= 5]

# --- Step 8: Reset index ---
df = df.reset_index(drop=True)

# --- Print cleaned data preview ---
print("🧹 Cleaned data preview (first 10 rows):")
print(df.head(10))
print("\n✅ Data cleaning complete!")
print(f"Rows remaining: {len(df)}\n")
print("-"*60 + "\n")

# --- Step 9: Save cleaned dataset ---
df.to_csv('cleaned_CN_EN_10K_Mylonas2017.csv', index=False)

# --- Step 10: Define broader color groups ---
color_groups = {
    'red': ['red', 'crimson', 'maroon', 'scarlet'],
    'orange': ['orange', 'amber', 'tangerine'],
    'yellow': ['yellow', 'gold', 'lemon'],
    'green': ['green', 'lime', 'olive', 'mint'],
    'blue': ['blue', 'navy', 'sky', 'azure', 'cyan'],
    'purple': ['purple', 'violet', 'lavender', 'magenta'],
    'pink': ['pink', 'rose', 'fuchsia'],
    'brown': ['brown', 'tan', 'beige', 'chocolate'],
    'grey': ['grey', 'gray', 'silver', 'charcoal'],
    'black': ['black'],
    'white': ['white', 'cream', 'ivory']
}

def map_to_group(name):
    for group, keywords in color_groups.items():
        for kw in keywords:
            if kw in name:
                return group
    return 'other'

df['color_group'] = df['name'].map(map_to_group)

# --- Step 11: Compute mean RGB per color group ---
grouped = df.groupby('color_group')[['red', 'green', 'blue']].mean().reset_index()

# --- Step 12: Convert mean RGB to hex for visualization ---
def rgb_to_hex(r, g, b):
    return '#{:02X}{:02X}{:02X}'.format(int(r), int(g), int(b))

grouped['hex_color'] = grouped.apply(lambda row: rgb_to_hex(row['red'], row['green'], row['blue']), axis=1)

# --- Step 13: Create Altair scatter plot with true colors ---
chart = alt.Chart(grouped).mark_circle(size=200).encode(
    x='red',
    y='green',
    color=alt.Color('hex_color:N', scale=None),
    tooltip=['color_group', 'red', 'green', 'blue']
).properties(
    width=800,
    height=600,
    title='Mean RGB Values per Color Group (True Colors)'
)

chart.show()
```

#### Insightful Description for Presentation

In this task, we transformed the raw colour naming dataset from *colornaming.net* into a **clean, structured, and analyzable format**. The workflow involved cleaning text entries, standardizing RGB values, removing invalid or rare entries, and grouping detailed colour names into broader categories. By computing **mean RGB values per colour group** and visualizing them in an interactive Altair chart, we ensured that each group’s colour representation is both **accurate and intuitive**.

This process highlights how **careful data preparation enhances both analytical reliability and visual interpretability**, creating a foundation for downstream analysis, machine learning, or geospatial visualization.


### Task 3: Informative Visualisations

The purpose of this task was to create **informative visualisations** that clearly communicate relationships and patterns in the cleaned colour naming dataset.  
This includes visualising the frequency of colour names and showing how these categories are distributed in RGB space.

#### Steps Performed

1. **Select the dataset:** Load the cleaned dataset from Task 2 (`cleaned_CN_EN_10K_Mylonas2017.csv`).  
2. **Aggregate data:** Count the frequency of each colour name and compute the **mean RGB values** for each name to summarise colour appearance.  
3. **Sort and format data:** Sort colour names by frequency to prioritise the most common colours, and ensure RGB values are numeric for visualisation.  
4. **Choose visualisation types:** Use a **bar chart** to display the top 20 most frequent colour names and a **scatter plot** to show the RGB distribution of the top 50 colours.  
5. **Apply colour encoding:** Map actual mean RGB values to the visualisations to represent colours accurately and intuitively.  
6. **Add interactivity:** Include tooltips showing colour name, frequency, and RGB values, enabling users to explore the dataset dynamically.  
7. **Combine visualisations:** Create a single interactive dashboard to present both frequency and RGB distribution together, enhancing clarity and user engagement.

This approach allows users to **quickly identify common colour names**, explore **colour similarity and variation**, and gain a deeper understanding of the dataset through **interactive visualisation**.

```python
import pandas as pd
# --- Step 1: Load the cleaned dataset ---
df = pd.read_csv('cleaned_CN_EN_10K_Mylonas2017.csv')

# --- Step 2: Aggregate data by colour name ---
freq_df = df.groupby('name').agg(
    count=('name', 'size'),
    mean_red=('red', 'mean'),
    mean_green=('green', 'mean'),
    mean_blue=('blue', 'mean')
).reset_index()

# --- Step 4: Convert mean RGB to hex for colour encoding ---
def rgb_to_hex(r, g, b):
    return '#{:02X}{:02X}{:02X}'.format(int(r), int(g), int(b))

freq_df['hex_color'] = freq_df.apply(
    lambda row: rgb_to_hex(row['mean_red'], row['mean_green'], row['mean_blue']), axis=1
)

# --- Step 5: Sort by frequency ---
freq_df = freq_df.sort_values(by='count', ascending=False)

# --- Step 6: Bar chart showing top 20 most frequent colour names ---
bar_chart = alt.Chart(freq_df.head(20)).mark_bar().encode(
    x=alt.X('count:Q', title='Frequency'),
    y=alt.Y('name:N', sort='-x', title='Colour Name'),
    color=alt.Color('hex_color:N', scale=None),
    tooltip=[
        alt.Tooltip('name:N', title='Colour Name'),
        alt.Tooltip('count:Q', title='Frequency'),
        alt.Tooltip('mean_red:Q', title='Mean Red'),
        alt.Tooltip('mean_green:Q', title='Mean Green'),
        alt.Tooltip('mean_blue:Q', title='Mean Blue')
    ]
).properties(
    width=800,
    height=600,
    title='Top 20 Most Frequent Colour Names'
)

# --- Step 6: Scatter plot for RGB distribution of top 50 colours ---
scatter_chart = alt.Chart(freq_df.head(50)).mark_circle(size=200).encode(
    x='mean_red:Q',
    y='mean_green:Q',
    color=alt.Color('hex_color:N', scale=None),
    tooltip=[
        alt.Tooltip('name:N', title='Colour Name'),
        alt.Tooltip('count:Q', title='Frequency'),
        alt.Tooltip('mean_red:Q', title='Mean Red'),
        alt.Tooltip('mean_green:Q', title='Mean Green'),
        alt.Tooltip('mean_blue:Q', title='Mean Blue')
    ]
).properties(
    width=800,
    height=600,
    title='RGB Distribution of Top 50 Colour Names'
)

# --- Step 7: Combine both charts into a single interactive dashboard ---
dashboard = alt.vconcat(bar_chart, scatter_chart).configure_title(
    fontSize=18,
    anchor='middle'
).interactive()

# --- Display dashboard ---
dashboard.show()
```

#### Insightful Description for Presentation

This visualisation provides a clear and interactive overview of the **most frequent colour names** and their **RGB distributions** in the dataset. The **horizontal bar chart** shows the frequency of the top 20 colour names, making it easy to compare which colours are most commonly used. Each bar is coloured using the **mean RGB values** for that colour, creating a direct link between the name and its visual appearance.

The **scatter plot** presents the RGB distribution of the top 50 colour names, allowing viewers to explore similarities and differences between colours. Colour encoding ensures that each point accurately represents the true appearance of the colour, while tooltips provide additional information such as frequency and exact RGB values.

This design emphasizes **clarity, interactivity, and informative insights**:

- **Top colours highlighted** to focus attention on the most meaningful data.  
- **Colour encoding with mean RGB values** provides an intuitive connection between data and perception.  
- **Tooltips** allow users to explore detailed information dynamically.  
- **Combined visualisations** give both frequency and RGB context, making patterns in colour usage and similarity immediately apparent.

Overall, this visualisation communicates both the **popularity of colour names** and their **real-world appearance**, enabling the audience to understand and explore the dataset in a visually engaging and insightful manner.


### Task 4: Misinformative Visualisations

The purpose of this task was to **create a visualisation that intentionally misleads** viewers about the frequency and characteristics of colours in the dataset. Using the cleaned colour naming dataset (`cleaned_CN_EN_10K_Mylonas2017.csv`), the visualisation combines selective data omission, arbitrary colour mapping, and shuffled ordering to distort perception while tooltips reveal the actual values.

#### Steps performed

1. **Load the cleaned dataset**: The CSV file was loaded and inspected to confirm its structure and contents.  
2. **Aggregate frequency of colour names**: Counts for each colour name were computed to understand their distribution.  
3. **Select a misleading subset**: The top 5 most frequent colours and 2 rare colours were chosen to exaggerate certain values and downplay others.  
4. **Compute mean RGB values for reference**: Mean red, green, and blue values were calculated for the selected colours.  
5. **Convert mean RGB to hex codes**: Hex codes were created for accurate colour swatches.  
6. **Generate fake RGB values for bars**: Arbitrary RGB values were assigned to the bars to mislead viewers about actual colour representation.  
7. **Create horizontal bar chart**: A bar chart shows the misleading frequency using fake colours, with interactive tooltips revealing real counts and RGB values.  
8. **Create RGB swatches in shuffled order**: Bars displaying true mean RGB colours were shuffled to further confuse perception.  
9. **Combine charts vertically**: The bar chart and swatches were stacked, maintaining interactivity and enhancing the misinformative effect.  

This workflow demonstrates how **design choices in scaling, colour, order, and selective data** can influence viewers’ perception, highlighting the importance of ethical and accurate visualisation practices.


```python
import pandas as pd
import altair as alt

# --- Step 1: Load the cleaned dataset ---
df = pd.read_csv('cleaned_CN_EN_10K_Mylonas2017.csv')

# --- Step 2: Aggregate frequency ---
color_counts = df['name'].value_counts().reset_index()
color_counts.columns = ['name', 'count']

# --- Step 3: Select subset to mislead (top 5 + 2 rare colours) ---
top5 = color_counts.head(5)
rare2 = color_counts.tail(2)
misleading_df = pd.concat([top5, rare2]).reset_index(drop=True)

# --- Step 4: Compute mean RGB values for reference ---
mean_rgb = df.groupby('name')[['red','green','blue']].mean().reset_index()
misleading_df = misleading_df.merge(mean_rgb, on='name', how='left')

# --- Step 5: Convert mean RGB to hex codes for swatches ---
def rgb_to_hex(r, g, b):
    return f'#{int(r):02X}{int(g):02X}{int(b):02X}'

misleading_df['true_hex'] = misleading_df.apply(lambda row: rgb_to_hex(row.red, row.green, row.blue), axis=1)

# --- Step 6: Generate fake RGB values for bars ---
misleading_df['fake_hex'] = misleading_df.index.map(
    lambda i: f'rgb({50+30*i},{200-20*i},{100+10*i})'
)

# --- Step 7: Create horizontal bar chart (misleading frequency) ---
bar_chart = alt.Chart(misleading_df).mark_bar().encode(
    x=alt.X('count:Q', title='Frequency (Misleading Scale)'),
    y=alt.Y('name:N', sort='-x', title='Colour Name'),
    color=alt.Color('fake_hex:N', scale=None, legend=None),
    tooltip=['name', 'count', 'red', 'green', 'blue', 'true_hex']
).properties(
    width=600,
    height=350,
    title='Misinformative Colour Frequency (Bars are misleading)'
)

# --- Step 8: Create RGB swatches in shuffled order ---
swatch_df = misleading_df.sample(frac=1)  # shuffle for misdirection

swatches = alt.Chart(swatch_df).mark_bar(height=30).encode(
    x=alt.X('name:N', title='Colour Name'),
    y=alt.value(1),
    color=alt.Color('true_hex:N', scale=None),
    tooltip=['name', 'red', 'green', 'blue', 'true_hex']
).properties(
    width=600,
    height=50,
    title='Actual Mean RGB Swatches (Reference)'
)

# --- Step 9: Combine charts vertically ---
final_chart = alt.vconcat(bar_chart, swatches).configure_axis(
    labelFontSize=12,
    titleFontSize=14
).configure_title(
    fontSize=16
)

final_chart.show()
```

#### Insightful Description for Presentation

This visualisation intentionally misleads the viewer about the distribution of colour names in the dataset, combining both quantitative and qualitative distortions. The horizontal bar chart shows only a subset of colours (top 5 most frequent plus 2 rare ones), making some colours appear more dominant than they actually are. Bars are coloured with arbitrary/fake RGB values, exaggerating differences and disconnecting the visual appearance from the actual data.

Below the bar chart, RGB swatches display the true mean colours, but are presented in a shuffled order, further confusing the audience and masking the true relationships between colours.

This design emphasizes distortion and misperception:

- **Selective data omission** exaggerates certain colour frequencies.  
- **Fake RGB values for bars** create visual bias and misrepresentation.  
- **Shuffled swatches and vertical layout** reduce clarity and make interpretation difficult.  

Overall, this visualisation demonstrates how intentional design choices—colour, scale, order, and selective data—can mislead viewers, underscoring the importance of ethical and transparent visualisation practices.


### Task 5: Categorical Colour Palette in Interactive Maps

This visualisation presents an **interactive map showing all colours from the dataset by country**, combining both **categorical clarity and geographic context**. Each country is represented with markers corresponding to the colour names present in the dataset, with marker size indicating the frequency of each colour. Using a **distinct categorical palette**, all colour names are easily distinguishable, ensuring clarity even with many categories.

**Interactive features enhance data exploration and user engagement:**  

- **Hovering** over markers reveals the country, colour name, and count.  
- The **legend includes all colour names**, allowing users to filter and focus on specific colours.  
- Users can zoom and pan to explore regions in more detail.  

This design emphasizes **clarity, completeness, and interactivity**:  

- Every colour from the dataset is represented in the legend to avoid omission.  
- Distinct categorical colours improve readability and help identify patterns across countries.  
- Interactive functionality allows users to explore complex data dynamically, providing a more engaging and insightful experience.  

Overall, this visualisation communicates **both the geographic distribution of colours and their relative frequencies**, offering an intuitive and interactive way to explore the dataset.


```python
import pandas as pd
import altair as alt
from vega_datasets import data

# --- Load dataset ---
df = pd.read_csv('cleaned_CN_EN_10K_Mylonas2017.csv')
country_counts = df['country_resident'].value_counts().reset_index()
country_counts.columns = ['code', 'count']

# Map codes to full names
country_name_map = {
    'US': 'United States of America',
    'GB': 'United Kingdom',
    'AU': 'Australia',
    'NZ': 'New Zealand',
    'IR': 'Iran'
}
country_counts['country_name'] = country_counts['code'].map(country_name_map)

# Assign colors
country_colors = {
    'United States of America': '#1f77b4',
    'United Kingdom': '#ff7f0e',
    'Australia': '#2ca02c',
    'New Zealand': '#d62728',
    'Iran': '#9467bd',
    'Other': 'lightgray'
}
country_counts['color'] = country_counts['country_name'].map(country_colors)

# --- Load TopoJSON and country codes ---
world_map = alt.topo_feature(data.world_110m.url, 'countries')
country_codes = pd.read_json('week6/world-110m-country-codes.json')
country_codes = country_codes[['id', 'name', 'code']]
country_codes['name'] = country_codes['name'].str.strip()
country_codes['code'] = country_codes['code'].str.strip()

# Merge dataset with country codes
country_counts_with_id = country_codes.merge(
    country_counts[['code', 'country_name', 'count', 'color']],
    on='code',
    how='left'
)

# Fill missing values
country_counts_with_id['country_name'] = country_counts_with_id['country_name'].fillna('Other')
country_counts_with_id['count'] = country_counts_with_id['count'].fillna(0)
country_counts_with_id['color'] = country_counts_with_id['color'].fillna('lightgray')

# Interactive selection
country_selection = alt.selection_point(fields=['country_name'], bind='legend')
```

```python
# --- Map chart ---
map_chart = alt.Chart(world_map).mark_geoshape(
    stroke='white',
    strokeWidth=0.5
).encode(
    color=alt.Color(
        'country_name:N',
        scale=alt.Scale(
            domain=list(country_colors.keys()),
            range=list(country_colors.values())
        ),
        legend=alt.Legend(
            title='Country',
            titleFontSize=13,
            labelFontSize=11,
            orient='right',
            padding=10,
            cornerRadius=5
        )
    ),
    tooltip=[
        alt.Tooltip('country_name:N', title='Country'),
        alt.Tooltip('code:N', title='Country Code'),
        alt.Tooltip('count:Q', title='Number of Entries', format=',')
    ],
    opacity=alt.condition(country_selection, alt.value(0.9), alt.value(0.6))
).transform_lookup(
    lookup='id',
    from_=alt.LookupData(country_counts_with_id, 'id', ['country_name', 'code', 'count', 'color'])
).add_params(
    country_selection
).project(
    type='naturalEarth1',
    scale=140,
    center=[5, 20]
).properties(
    width=900,
    height=520,
    title='Colour Naming Dataset: Participant Distribution by Country'
)

map_chart
```

```python
# --- Bar chart ---
bar_chart = alt.Chart(country_counts).mark_bar().encode(
    x=alt.X('country_name:N', title='Country', axis=alt.Axis(labelAngle=-45), sort='-y'),
    y=alt.Y('count:Q', title='Number of Entries'),
    color=alt.Color('country_name:N',
        scale=alt.Scale(domain=list(country_colors.keys()), range=list(country_colors.values())),
        legend=None
    ),
    tooltip=[
        alt.Tooltip('country_name:N', title='Country'),
        alt.Tooltip('count:Q', title='Entries', format=',')
    ],
    opacity=alt.condition(country_selection, alt.value(0.9), alt.value(0.6))
).add_params(
    country_selection
).properties(
    width=450,
    height=400,
    title='Entries per Country'
)
# --- Pie chart ---
pie_chart = alt.Chart(country_counts).mark_arc().encode(
    theta=alt.Theta('count:Q', title='Entries'),
    color=alt.Color('country_name:N',
        scale=alt.Scale(domain=list(country_colors.keys()), range=list(country_colors.values())),
        legend=None
    ),
    tooltip=[
        alt.Tooltip('country_name:N', title='Country'),
        alt.Tooltip('count:Q', title='Entries', format=',')
    ],
    opacity=alt.condition(country_selection, alt.value(0.9), alt.value(0.6))
).add_params(
    country_selection
).properties(
    width=450,
    height=450,
    title='Proportion of Entries by Country'
)
charts = alt.hconcat(bar_chart, pie_chart).resolve_scale(color='shared')
charts
```

#### Interactive Map: Categorical Colour Palette and User Engagement

This visualisation presents an **interactive map** showing the distribution of colour naming dataset entries by country, combining **categorical clarity** and **geographic context**. Each country is represented with a distinct, easily distinguishable colour, while the frequency of dataset entries is reflected in linked **bar and pie charts**.

#### Interactive Features and Their Contribution to User Engagement

The interactive map incorporates several features designed to enhance user engagement and facilitate in-depth data exploration:

1. **Hover Tooltips:**  
   When users hover over a country, a tooltip displays detailed information, including the country name, ISO code, and the number of dataset entries. This allows users to quickly gain insights without cluttering the visualisation.

2. **Legend-Based Selection:**  
   The legend lists all countries with their corresponding colours. Users can select countries in the legend to highlight them on the map and in associated charts (bar and pie charts). This enables focused comparisons and filtering of specific categories.

3. **Opacity Highlighting:**  
   Selected countries appear more prominent, while unselected countries are semi-transparent. This visual cue draws attention to areas of interest and helps users identify patterns or outliers more easily.

4. **Interactive Linking Across Charts:**  
   The selection on the map dynamically affects the bar chart and pie chart, so users can explore relationships between countries dataset counts and proportions. This coordinated interaction supports multi-dimensional data exploration.

### Contribution to Data Exploration

These features collectively improve the user's ability to understand complex datasets by:

By integrating these interactive features, the visualization encourages users to explore, analyze, and interpret the dataset effectively, turning a static map into a dynamic tool for insight discovery.
